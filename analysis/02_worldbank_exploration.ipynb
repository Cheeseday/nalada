```python
indicators = {
    'SP.URB.TOTL.IN.ZS': 'urban_population_pct',
    'SP.URB.GROW':        'urban_growth_rate',
    'EN.POP.DNST':        'population_density',
    'SP.URB.TOTL':        'urban_population_total'
}

for code, name in indicators.items():
    df = wb.data.DataFrame(code)
    print(f"\n{name} ({code})")
    print(f"  Shape: {df.shape}")
    print(f"  Index type: {df.index.name}")
    print(f"  Columns sample: {list(df.columns[:5])}")
```

```python
extra_indicators = {
    'EN.ATM.PM25.MC.M3': 'pm25_exposure',
    'EN.POP.SLUM.UR.ZS': 'slum_population_pct',
    'EG.ELC.ACCS.UR.ZS': 'electricity_access_pct',
    'AG.LND.TOTL.UR.K2': 'urban_land_area_km2'
}

for code, name in extra_indicators.items():
    df = wb.data.DataFrame(code)
    non_null = df.notna().sum().sum()
    print(f"{name}: shape={df.shape}, non-null values={non_null}")
```
