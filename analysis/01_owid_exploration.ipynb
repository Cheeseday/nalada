```python
import pandas as pd

url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)
```

```python
# Shape and columns
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print("Columns:", df.columns.tolist())
```

```python
# Year range
print(f"Years: {df['year'].min()} - {df['year'].max()}")
```

```python
# Not a countries
non_countries = df[df['iso_code'].isna()]['country'].unique()
print("Non-country rows:", non_countries)
```

```python
# NULL % for key decoupling columns specifically
key_cols = [
    'co2', 'co2_per_capita', 'co2_per_gdp',
    'consumption_co2', 'consumption_co2_per_gdp',
    'trade_co2', 'trade_co2_share',
    'gdp', 'population',
    'temperature_change_from_co2',
    'share_of_temperature_change_from_ghg',
    'renewables_share_energy'
]

# Filter only existing columns
key_cols = [c for c in key_cols if c in df.columns]
null_pct = (df[key_cols].isnull().sum() / len(df) * 100).round(1)
print(null_pct.sort_values(ascending=False))
```

```python
# Year range for key columns - when does consumption_co2 data actually start?
for col in ['co2', 'consumption_co2', 'trade_co2', 'gdp']:
    if col in df.columns:
        valid = df[df[col].notna()]
        print(f"{col}: {int(valid['year'].min())} - {int(valid['year'].max())}")
```

```python
# Spot check
key_cols_check = ['year', 'co2', 'co2_per_capita', 'consumption_co2', 'trade_co2', 'population']
print(df[df['country'] == 'Germany'][key_cols_check].tail(10).to_string())
```
